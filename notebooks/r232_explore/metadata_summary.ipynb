```python
# Import statements
import pandas as pd
```

```python
#File locations
bac_meta = "../../r232/bac120_metadata_r232.tsv"
bac_tax = "../../r232/bac120_taxonomy_r232.tsv"
bac_tree = "../../r232/bac120_tree_r232.tsv"

archea_meta = "../../r232/ar53_metadata_r232.tsv"
archea_tax = "../../r232/ar53_taxonomy_r232.tsv"
archea_tree = "../../r232/ar53_tree_r232.tsv"
```

## Metadata table processing

```python
bac_meta_df = pd.read_csv(bac_meta, sep="\t")
bac_meta_df["domain"] = "Bacteria"
archea_meta_df = pd.read_csv(archea_meta, sep="\t")
archea_meta_df["domain"] = "Archaea"
stacked_meta_df = pd.concat([bac_meta_df, archea_meta_df], axis=0)

stacked_meta_df
```

```python
## Unique accessions

print(f"Number of unique accessions: {stacked_meta_df['accession'].nunique()}")
print(f"Number of rows: {stacked_meta_df.shape[0]}")
print(f"is the number of unique accessions equal to the number of rows? {stacked_meta_df['accession'].nunique() == stacked_meta_df.shape[0]}")
```

Make sense as the accession ID is effectively the index for the metadata tables.

## Taxonomy table

```python
bac_tax_df = pd.read_csv(bac_tax, sep="\t", header=None)
bac_tax_df["domain"] = "Bacteria"
archea_tax_df = pd.read_csv(archea_tax, sep="\t", header=None)
archea_tax_df["domain"] = "Archaea"
stacked_tax_df = pd.concat([bac_tax_df, archea_tax_df], axis=0)
header = ["accession", "taxonomy", "domain"]
stacked_tax_df.columns = header

stacked_tax_df
```

```python
# accessions per taxa 

unique_taxa = stacked_tax_df["taxonomy"].nunique()
unique_accessions = stacked_tax_df["accession"].nunique()
print(f"Number of unique taxa: {unique_taxa}")
print(f"Number of unique accessions: {unique_accessions}")
print(f"Is the number of unique taxa equal to the number of unique accessions? {unique_taxa == unique_accessions}")
```

Some taxa must be represented by multiple species

```python
accessions_per_taxa = stacked_tax_df.groupby("taxonomy")["accession"].nunique()

accessions_per_taxa.describe()
```

Most species have 1 a few seem to have a lot

```python
top_ten = accessions_per_taxa.sort_values(ascending=False).head(10)
top_ten
```

These seem to be the most common species.
