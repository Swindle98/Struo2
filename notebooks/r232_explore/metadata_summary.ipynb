```python
split_tree_filtered_reps = filtered_reps.loc[:,["accession", "ncbi_organism_name", "gtdb_taxonomy"]]
tax_levels = ["Domain", "Phylum", "Class", "Order", "Family", "Genus", "Species"]
split_tree_filtered_reps[tax_levels] = split_tree_filtered_reps["gtdb_taxonomy"].str.split(";", expand=True)
split_tree_filtered_reps = split_tree_filtered_reps.drop(columns="gtdb_taxonomy", axis=1)

split_tree_filtered_reps.head(10)
```

```python
from skbio.tree import TreeNode

# Helper function 

def summarize_tree(tree: TreeNode) -> dict:
    """
    Print and return summary stats for a skbio TreeNode.
    Computes tip depths in a single traversal rather than calling
    tree.distance() per tip, which is much faster on large trees.
    """
    n_tips = tree.count(tips=True)
    n_nodes = tree.count(tips=False)
    n_internal = n_nodes - n_tips

    # Branch lengths across all nodes
    lengths = [n.length for n in tree.traverse() if n.length is not None]
    total_branch_length = sum(lengths)
    n_missing_lengths = sum(1 for n in tree.traverse() if n.length is None)

    # Root-to-tip depths via single pass, accumulating from parent depths
    depths = {}
    tip_depths = []
    for node in tree.preorder(include_self=True):
        if node.parent is None:
            depths[node] = 0.0
        else:
            parent_depth = depths[node.parent]
            branch_len = node.length if node.length is not None else 0.0
            depths[node] = parent_depth + branch_len
        if node.is_tip():
            tip_depths.append(depths[node])

    stats = {
        "n_tips": n_tips,
        "n_internal_nodes": n_internal,
        "total_branch_length": total_branch_length,
        "n_missing_branch_lengths": n_missing_lengths,
        "max_tip_depth": max(tip_depths),
        "min_tip_depth": min(tip_depths),
        "mean_tip_depth": sum(tip_depths) / len(tip_depths),
    }

    print(f"Tips: {stats['n_tips']}")
    print(f"Internal nodes: {stats['n_internal_nodes']}")
    print(f"Total branch length: {stats['total_branch_length']:.2f}")
    print(f"Nodes missing branch length: {stats['n_missing_branch_lengths']}")
    print(f"Max tip depth: {stats['max_tip_depth']:.4f}")
    print(f"Min tip depth: {stats['min_tip_depth']:.4f}")
    print(f"Mean tip depth: {stats['mean_tip_depth']:.4f}")

    return stats
```

```python
tip_names = [tip.name.replace(' ', '_') for tip in bac_tree.tips()]
print(tip_names[60:80])  # Print the first 10 tip names for verification
bac_meta_tree_filtered = bac_meta_df[["accession", "gtdb_taxonomy"]]
bac_meta_tree_filtered[tax_levels] = bac_meta_tree_filtered["gtdb_taxonomy"].str.split(";", expand=True)
bac_meta_tree_filtered = bac_meta_tree_filtered.drop(columns="gtdb_taxonomy", axis=1)
bac_meta_tree_filtered = bac_meta_tree_filtered.loc[bac_meta_tree_filtered["accession"].isin(tip_names)]
bac_meta_tree_filtered.info()
```

```python
combined_meta_tree_filtered = pd.concat([arc_meta_tree_filtered, bac_meta_tree_filtered], axis=0)
combined_meta_tree_filtered.info()
```

```python
print(f"Number of unique species in combined metadata tree: {combined_meta_tree_filtered['Species'].nunique()}")
print(f"Number of rows in combined metadata tree: {combined_meta_tree_filtered.shape[0]}")
```

```python
# Binding the trees at root

combined_tree = TreeNode(name="root")

combined_tree.extend([bac_tree, arc_tree])

summarize_tree(combined_tree)
```

```python
# Renaming the tips with species

name_mapping = dict(zip(combined_meta_tree_filtered["accession"], combined_meta_tree_filtered["Species"]))
unmatched_tips = []
for tip in combined_tree.tips():
    underscored_name = tip.name.replace(' ', '_')
    if underscored_name in name_mapping:
        tip.name = name_mapping[underscored_name]
    else:
        unmatched_tips.append(tip.name)

print(f"Number of unmatched tips: {len(unmatched_tips)}")
```

```python
tip_names = {tip.name for tip in combined_tree.tips()}
species_names = set(split_tree_filtered_reps['Species'])

intersection = tip_names & species_names
only_in_meta = species_names - tip_names
only_in_tree = tip_names - species_names
union = tip_names | species_names

print(f"intersect(tips, species): {len(intersection)}")
print(f"setdiff(meta, tree):      {len(only_in_meta)}")
print(f"setdiff(tree, meta):      {len(only_in_tree)}")
print(f"union(tips, species):     {len(union)}")
```

```python
# prune and keep only whats in the filtered metadata

tree_pruned = combined_tree.shear(intersection)

summarize_tree(tree_pruned)
```

```python
tip_names = {tip.name for tip in tree_pruned.tips()}
species_names = set(split_tree_filtered_reps['Species'])

intersection = tip_names & species_names
only_in_meta = species_names - tip_names
only_in_tree = tip_names - species_names
union = tip_names | species_names

print(f"intersect(tips, species): {len(intersection)}")
print(f"setdiff(meta, tree):      {len(only_in_meta)}")
print(f"setdiff(tree, meta):      {len(only_in_tree)}")
print(f"union(tips, species):     {len(union)}")
```

```python
# Save tree

tree_pruned.write("../../r232/pruned_tree.tree", format="newick")
```
